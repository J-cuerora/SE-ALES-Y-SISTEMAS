# LISTA CICLO **FOR**

```
# Lista original
lista = [1, "andres", 8, 9, 2]

# Lista vacía donde guardaremos los elementos invertidos
invertida = []

# Ciclo for recorriendo la lista de atrás hacia adelante
for i in range(len(lista) - 1, -1, -1):
    invertida.append(lista[i])

print("Lista original:", lista)
print("Lista invertida:", invertida)
```

# **LISTA CICLO WHILE**

```
lista = [1, "andres", 8, 9, 2]
invertida = []
i = len(lista)-1
while i >= 0:
    invertida.append(lista[i])
    i -= 1
print("Lista original:", lista)
print("Lista invertida:", invertida)
```

# REPASO DE CLASES

```
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        return f"Hola, mi nombre es {self.nombre} y tengo {self.edad} años."

# Crear objeto
persona1 = Persona("Ana", 25)

# Usar métodos
print(persona1.saludar())
```

# CLASES HERENCIA

```
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def describir(self):
        return f"{self.nombre}, {self.edad} años"


# Clase hija que hereda de Persona
class Estudiante(Persona):
    def __init__(self, nombre, edad, carrera):
      super().__init__(nombre, edad)
      self.carrera = carrera

    def describir_estudiante(self):
        return f"{self.describir()} - Estudia {self.carrera}"


# Crear objeto de la clase hija
estudiante1 = Estudiante("Luis", 20, "Ingeniería")

print(estudiante1.describir_estudiante())
```

```
class Animal:
    def sonido(self):
        return "Sonido genérico"

class Perro(Animal):
    def sonido(self):
        return "Guau guau"  # Sobrescritura del método

class Gato(Animal):
    def sonido(self):
        return "Miau"

# Probar
animal = Animal()
perro = Perro()
gato = Gato()

print(animal.sonido())
print(perro.sonido())
print(gato.sonido())
```
