# Transformada de Fourier

 - Se presentan los conceptos básicos de la transformada de Fourier, en sus versiones continua, tiempo discreto y transformada discreta (extendida a su versión rápida).

- La transformada de Fourier se puede entender como la extensión de la serie de Fourier para $t\in(-\infty,+\infty)$:

\begin{equation}
X(\omega) = \mathscr{F} \{x(t) \} = \int_{-\infty}^{+\infty}{x(t)e^{-j\omega t}dt}.
\end{equation}

- La respectiva transformada inversa (extensión de la reconstrucción mediante serie de Fourier con $t\in(-\infty,+\infty)$) se define como:

\begin{equation}
x(t) = \mathscr{F}^{-1} \{X(\omega) \} = \frac{1}{2\pi}\int_{-\infty}^{+\infty}{X(\omega)e^{j\omega t}d\omega}.
\end{equation}

- Cabe resaltar entonces que el concepto de espectro en frecuencia para la transformada de Fourier asume que $\omega \in (-\infty,+\infty).$

- En resumen, la serie de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento finito $t\in [-T/2,T/2]$ o $t\in [t_i,t_f]$, en espectro en frecuencia discreto $c_n \in \mathbb{R},\mathbb{C}$, con $n\in \mathbb{Z}.$

- La transformada de Fourier analiza $x(t)\in \mathbb{R},\mathbb{C}$ en tiempo continuo sobre segmento infinito de tiempo $t\in (-\infty,+\infty)$, en espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R},\mathbb{C}$, con $\omega\in (-\infty,+\infty).$


- Las propiedades principales de la transformada de Fourier son útiles para analizar el espectro en fenómenos físicos clásicos (Ver material apoyo [propiedades de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Propiedades_TransformadaFourier.pdf)).

- Además, algunas de las transformadas básicas se pueden encontrar en las [Tablas de la transformada de Fourier](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/Tablas_TransformadaFourier.pdf).

# Transformada de Fourier en tiempo discreto

- A partir de la transformada de Fourier, se puede analizar también una señal en tiempo discreto $x[k]\in \mathbb{R} ,\mathbb{C},$ con $k\in\mathbb{Z}$, con el fin de calcular su espectro en frecuencia continuo (función de densidad espectral) $X(\omega)\in \mathbb{R} ,\mathbb{C}$, pero con $\omega$ continua en el intervalo $\omega\in [-\pi,+\pi]$ o $\omega\in [0,2\pi],$ dado que las bases de representación requieren restringir la codificación de la señal $x[k]$ utilizando funciones seno y coseno originales según la expresión:

\begin{equation}
X(\omega) = \sum_{k\in\mathbb{Z}}{x[k]e^{-j\omega k}},
\end{equation}

- **Nota**: Recuerde que cos/sen en tiempo discreto presentan periodicidad de $2\pi$ en la frecuencia angular $\cos[(\omega\pm r 2\pi)k]=\cos[\omega k]$; con $r,k \in \mathbb{Z}$.

- A la expresión anterior se le conoce comunmente como la trasformada de Fouirer en tiempo discreto (Discrete time Fourier transform - DTFT).
y su respectiva transformada inversa (iDTFT) se expresa como:


\begin{equation}
x[k] = \frac{1}{2\pi}\int_{-\pi}^{+\pi}{X(\omega)e^{j\omega k}d\omega}=\frac{1}{2\pi}\int_{0}^{+2\pi}{X(\omega)e^{j\omega k}d\omega}
\end{equation}


# Transformada discreta de Fourier y transformada rápida de Fourier

- Sin embargo, tanto para la serie de Fourier, como para la transformada (continua y discreta en tiempo), se require de la resolución analítica de integrales; además, de la suposición de intervalos infinitos en tiempo y frecuencia (para el caso de la transformada de Fourier), los cuales no son viables ni implementables en la práctica.

- Por consiguiente, el concepto de espectro en frecuencia se extiende a sistemas de proceso de digital (tiempo discreto y  frecuencia discreta), mediante el cálculo de la transformada discreta de Fourier (Discrete Fourier Transform - DFT).

- En este caso, la transformada discreta se define como:

\begin{equation}
X[n] = \sum_{k=0}^{K-1}{x[k]e^{-jk\frac{2\pi n}{K}}},
\end{equation}

**Nota**: observe que el ángulo en la función euler $\theta_n = \frac{k 2\pi n}{K}$se restringe al intervalo $[0,2\pi)$ para $k\in\{0,1,\dots,K-1\}$, dado que se trabaja con funciones cos/sen en tiempo discreto.

- La DFT es una discretización en frecuencia de la DTFT:

\begin{equation}
X[n] = X(\omega)\Big|_{\omega=\frac{2\pi n}{K}},
\end{equation}

- Respecto a la serie de Fourier, la DFT se puede relacionar mediante:

\begin{equation}
X[n] = K c_n,
\end{equation}
asumiendo señal en tiempo discreto y $\omega_o = \frac{2\pi}{K}$ en representación por serie.

- Por otra parte, la transforma discreta de Fourier inversa (iDFT) se define como:

\begin{equation}
x[k] = \sum_{n=0}^{K-1}{X[n]e^{jk\frac{2\pi n}{K}}}.
\end{equation}

Nuevamente, note que las variables tiempo y frecuencia discretas $n,k\in\{0,1,\dots,K-1\}.$ Dado que la frecuencia fundamental se restringe como $\omega_o=\frac{2\pi}{K},$ las bases de representación en tiempo discreto codificaran frecuencias angulares en el rango $[0,2\pi)$ (cosenos y senos originales).

- No obstante, en la práctica el cálculo de las sumatorias es costoso computacionalmente, a medida que $K$ aumenta. Por consiguiente, se cuentan con paquetes computacionales que optimizan el cálculo de la DFT de forma rápida, lo que conlleva a la comunmente conocida transformada rápida de Fourier ([Fast Fourier Transform - FFT](https://es.wikipedia.org/wiki/Transformada_r%C3%A1pida_de_Fourier)).

# Ejemplo:

- Se presenta el cálculo del espectro mediante DFT (`fft` de numpy) para la señal:


\begin{equation}
x(t) = \begin{cases} A, & -T/2 \leq t < 0 \\ -A, & 0 \leq t < T/2 \end{cases}
\end{equation}

```python
# Para graficar
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import warnings
warnings.filterwarnings("ignore")

mpl.rc('axes', labelsize=14)
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)

A = 1 # amplitud de la señal
T = 5 # segundos de gráfica
Fs = 1000 # frecuencia de muestreo definida
tv = np.arange(-T/2,T/2,1/Fs) #vector de tiempo generado
Nm = len(tv) #numero de muestras requeridas

#se crea la señal A y -A concatenando
xo = np.concatenate((A*np.ones(int(Nm/2)), -A*np.ones(int(Nm/2))), axis=0)
#graficar señal original
plt.plot(tv,xo,'r',linewidth=4)
plt.grid()
plt.xlabel("t[s]",fontsize = 14)
plt.ylabel("x(t)",fontsize = 14)
plt.show()
```

```python
#cálculo de la FFT de la señal signo invertido
vf = np.fft.fftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.fft(xo) #cálculo FFT

plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/len(vf),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.show()
```

**Nota**: La función `np.fft.fftfreq` cálcula el equivalente en Hz de cada armónico, teniendo en cuenta que en tiempo discreto:

$$\theta_n = \frac{k 2\pi n}{K} = k \omega_n = k n \omega_o$$

Además, en el proceso de discretización ($F_s$ frecuencia de muestreo):

$$\omega_n t = 2\pi F_n t =  \frac{2\pi F_n k}{F_s} = 2 \pi f n k = \frac{2 \pi n k}{K}$$

donde $f=\frac{1}{K}$ es equivalente a la frecuencia en Hz en tiempo discreto.

Así:

$$\frac{2\pi F_n k}{F_s}=\frac{2 \pi n k}{K}$$

y:

$$F_n = \frac{F_s n}{K} [Hz]$$




- Note que:

$x(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega$

$x^*(t) = \left( \frac{1}{2\pi}\int_{-\infty}^\infty X(\omega) e^{j\omega t} d\omega\right)^*=\frac{1}{2\pi}\int_{-\infty}^\infty X^*(\omega) e^{-j\omega t} d\omega$

Haciendo $\omega' = - \omega$:


$x^*(t) = \frac{1}{2\pi}\int_{-\infty}^\infty X^*(-\omega') e^{j\omega' t} d\omega'$

Entonces:

$\mathscr{F}\{x^*(t)\} = X^*(-\omega) = \mathscr{Re}\{X(-\omega) \} -j\mathscr{Im}\{X(-\omega) \} $

Y

$\mathscr{F}\{x(t)\} = X(\omega) = \mathscr{Re}\{X(\omega) \} +j\mathscr{Im}\{X(\omega) \} $

Por lo tanto:

$\mathscr{Re}\{X(\omega) \}=\mathscr{Re}\{X(-\omega) \}$
$\mathscr{Im}\{X(\omega) \}=-\mathscr{Im}\{X(-\omega) \}$

-Dado que el espectro es simétrico para señales reales ($x(t)\in \mathbb{R}$), se sugiere utilizar la expresión simplificada `rfft`:

```python
vf = np.fft.rfftfreq(len(xo),1/Fs) #calculo vector de frecuencias
Xw = np.fft.rfft(xo) #cálculo FFT
plt.stem(vf,abs(Xw),'r')
plt.title(r'|X[n]|')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|$',fontsize = 14)
plt.axis('tight')
plt.grid()
plt.show()
plt.stem(vf,abs(Xw)/(2*len(vf)),'r')
plt.title(r'|X[n]|/K=c_n')
plt.xlabel(r'$f$[Hz]',fontsize = 14)
plt.ylabel(r'$|X[n]|/K$',fontsize = 14)
plt.grid()
plt.show()
```

- La reconstrucción de x[k] se puede obtener mediante la transformada inversa:

```python
xe = np.fft.irfft(Xw) #transformada inversa
plt.plot(tv,xe)
plt.title(r'Reconstrucción completa')
plt.xlabel(r't',fontsize = 14)
plt.ylabel(r'$x[n]$',fontsize = 14)
plt.show()
```

- Se pueden escoger o ponderar los coeficientes para filtrar (alterar) la reconstrucción de la señal.

```python
Xwf = Xw.copy()
Xwf[10:] = 0 #solo se tienen en cuenta los 10 primeros armónicos
xe2 = np.fft.irfft(Xwf)
plt.plot(tv,xe2)
plt.show()
```

# Tipos de filtrado

- Según el rango de frecuencias a filtrar (según el diagrama de magnitud), los filtros se pueden clasificar en:

  - Filtro pasa-bajas (low-pass)
  - Filtro pasa-altas (high-pass)
  - Filtro pasa-banda (band-pass)
  - Filtro rechaza banda (band-stop)

  ![filtros](https://github.com/amalvarezme/SenalesSistemas/blob/master/3_SerieyTransformadaFourier/filters2.jpg?raw=1)





- En este caso, se plantea un filtro pasa banda, apagando los armónicos que no nos interesa.

```python
#filtrar espectro
Xwf = Xw.copy()
f1 = 100 #frecuencia en Hz corte 1
f2 = 800 #frecuencia en Hz corte 2
ind = ~((vf > f1) & (vf < f2)) #frecuencias eliminar-> recueder que ~ actua como negación
Xwf[ind,:] = 0
plt.plot(vf,abs(Xwf))
plt.legend(('canal 1','canal 2'))
plt.show()
```

```python
xe2 = np.fft.irfft(Xwf,axis=0) #fft inversa sobre los dos canales de audio
```

El modelo que utiliza la **Transformada Rápida de Fourier (FFT)** para el cálculo de armónicos se basa fundamentalmente en la **Transformada Discreta de Fourier (DFT)** y el principio del **Análisis de Fourier**.

## Modelo de la FFT para el Cálculo de Armónicos

La FFT es un **algoritmo optimizado** que calcula de manera eficiente la DFT. Su propósito central es **transformar una señal del dominio del tiempo al dominio de la frecuencia**. En el contexto del análisis de armónicos, este modelo permite **descomponer** una señal periódica no sinusoidal (como una onda de voltaje o corriente distorsionada) en una suma de sus componentes sinusoidales individuales: la **frecuencia fundamental** y sus **armónicos**.

### 1. Principio Fundamental: Análisis de Fourier

El modelo parte del **Teorema de Fourier**, que establece que cualquier señal periódica puede representarse como una suma (serie) de funciones sinusoidales (senos y cosenos) cuyas frecuencias son múltiplos enteros de la **frecuencia fundamental** ($f_1$). Estas frecuencias múltiples son los **armónicos** ($h \cdot f_1$, donde $h$ es el orden del armónico).

La señal periódica $f(t)$ se representa idealmente mediante la **Serie de Fourier** como:

$$f(t) = A_0 + \sum_{h=1}^{\infty} A_h \cdot \sin(2\pi h f_1 t + \phi_h)$$

Donde:
* $A_0$: Componente de corriente continua (DC).
* $A_h$: Amplitud (magnitud) del armónico de orden $h$.
* $f_1$: Frecuencia fundamental (ej. 50 o 60 Hz en sistemas eléctricos).
* $\phi_h$: Fase del armónico de orden $h$.

### 2. Implementación Digital: La DFT y la FFT

Dado que en el mundo real se trabaja con señales muestreadas y discretas, el modelo utiliza:

* **Transformada Discreta de Fourier (DFT):** Es la herramienta matemática que convierte una secuencia finita de muestras de tiempo, $x[n]$, en una secuencia finita de muestras de frecuencia, $X[k]$. Cada valor $X[k]$ representa la magnitud y fase de una componente sinusoidal específica de la señal.
    $$X[k] = \sum_{n=0}^{N-1} x[n] \cdot e^{-j 2 \pi k n / N}$$
    Donde $N$ es el número total de muestras y $k$ es el índice de frecuencia (relacionado con el orden armónico).

* **Transformada Rápida de Fourier (FFT):** Es simplemente una **implementación algorítmica** de la DFT que reduce drásticamente el número de cálculos, haciéndola factible para el procesamiento en tiempo real o casi real. La eficiencia de la FFT se obtiene típicamente cuando el número de muestras ($N$) es una potencia de 2.

### 3. Descripción del Proceso de Cálculo de Armónicos

El proceso para el análisis de armónicos usando la FFT sigue estos pasos:

1.  **Adquisición de la Señal (Muestreo):** La señal de voltaje o corriente en el dominio del tiempo ($f(t)$) es muestreada a una **frecuencia de muestreo** ($f_s$) lo suficientemente alta (mínimo, el doble de la máxima frecuencia armónica de interés, según el **Teorema de Nyquist**). Esto produce una secuencia de datos discreta, $x[n]$.
    * *Importante:* Para una correcta detección de armónicos, se debe capturar un número exacto de períodos de la frecuencia fundamental para minimizar el **efecto de fuga espectral** (spectral leakage).

2.  **Aplicación del Algoritmo FFT:** Se aplica el algoritmo FFT a la secuencia de muestras $x[n]$ (posiblemente después de aplicar una **función ventana** para mitigar la fuga espectral). El resultado es el **espectro de frecuencia complejo** $X[k]$.

3.  **Obtención del Espectro de Magnitud y Fase:** La salida compleja $X[k]$ contiene la información de **magnitud** y **fase** para cada componente de frecuencia.
    * La **magnitud** de $X[k]$ es proporcional a la amplitud ($A_h$) de la onda sinusoidal en la frecuencia correspondiente.
    * La **fase** de $X[k]$ es el ángulo de fase ($\phi_h$) de esa componente.

4.  **Identificación de Armónicos:** La resolución de frecuencia ($\Delta f$) de la FFT está dada por $\Delta f = f_s / N$. Los picos en el espectro de magnitud que ocurren en múltiplos enteros de la frecuencia fundamental ($h \cdot f_1$) se identifican como los **armónicos**. Por ejemplo, para $f_1 = 60\text{ Hz}$, se buscarían picos en $120\text{ Hz}$ (2do armónico), $180\text{ Hz}$ (3er armónico), y así sucesivamente.

5.  **Cálculo de Parámetros:** Con las amplitudes de los armónicos ($A_h$) y la fundamental ($A_1$), se pueden calcular métricas como:
    * **Amplitud de cada Armónico:** La magnitud del pico en la frecuencia armónica.
    * **Distorsión Armónica Total (THD):** Mide la cantidad de distorsión presente, comparando la potencia de los armónicos con la potencia de la fundamental.

En resumen, la FFT permite a los ingenieros **cuantificar** la distorsión de una señal al aislar y medir la magnitud y fase de la frecuencia fundamental y de cada uno de sus armónicos.

```python
import numpy as np
import time
import matplotlib.pyplot as plt

# =================================================================
# 1. Definición de la DFT Directa (O(N^2))
# =================================================================
def dft_directa(x):
    """Calcula la Transformada Discreta de Fourier (DFT) directamente."""
    N = len(x)
    X = np.zeros(N, dtype=complex)

    # Se calcula la componente X[k] para cada k de 0 a N-1
    for k in range(N):
        sumatoria = 0.0 + 0.0j
        # Se realiza la sumatoria de n de 0 a N-1
        for n in range(N):
            # Exponencial compleja W_N^(kn)
            angulo = -2 * np.pi * k * n / N
            sumatoria += x[n] * np.exp(1j * angulo)
        X[k] = sumatoria
    return X

# =================================================================
# 2. Definición de la FFT (O(N log N))
# =================================================================
def fft_numpy(x):
    """Calcula la Transformada Rápida de Fourier (FFT) usando NumPy."""
    return np.fft.fft(x)

# =================================================================
# 3. Pruebas y Medición de Tiempos
# =================================================================

# Longitudes de señal solicitadas (potencias de 2)
N_values = [4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096]

dft_times = []
fft_times = []

print("Iniciando medición de tiempos...")

for N in N_values:
    # 1. Crear señal de prueba de longitud N (e.g., ruido aleatorio)
    x = np.random.rand(N) + 1j * np.random.rand(N)

    # 2. Medir tiempo de DFT Directa (Solo para N <= 1024, para evitar tiempos excesivos)
    dft_time = 0.0
    if N <= 1024:
        start_time = time.time()
        _ = dft_directa(x)
        end_time = time.time()
        dft_time = end_time - start_time
    dft_times.append(dft_time)

    # 3. Medir tiempo de FFT
    start_time = time.time()
    _ = fft_numpy(x)
    end_time = time.time()
    fft_time = end_time - start_time
    fft_times.append(fft_time)

    print(f"N={N}: DFT_Directa={dft_time:.6f} s, FFT={fft_time:.6f} s")

print("Medición de tiempos completada.")

# =================================================================
# 4. Generación de Gráfico Comparativo
# =================================================================

# Filtrar resultados de DFT para que no se muestren ceros si se omitieron cálculos
N_dft_plot = [N_values[i] for i, t in enumerate(dft_times) if t > 0]
dft_times_plot = [t for t in dft_times if t > 0]

plt.figure(figsize=(10, 6))
# Gráfica para la DFT Directa (Escala O(N^2))
plt.plot(N_dft_plot, dft_times_plot, 'o-', label='DFT Directa ($O(N^2)$)', color='red')
# Gráfica para la FFT (Escala O(N log N))
plt.plot(N_values, fft_times, 'x-', label='FFT (NumPy) ($O(N \log N)$)', color='blue')

# Configuración del gráfico
plt.xlabel('Longitud de la Señal de Entrada (N)')
plt.ylabel('Tiempo de Ejecución (Segundos)')
plt.title('Comparación de Tiempo de Cómputo: DFT Directa vs FFT')
plt.xscale('log') # Escala logarítmica para N, para mostrar mejor el crecimiento
plt.yscale('log') # Escala logarítmica para el tiempo, para apreciar la diferencia
plt.grid(True, which="both", ls="--")
plt.legend()
plt.tight_layout()
plt.show()
```

La **frecuencia de $-3\text{ dB}$** es la **definición estándar** y el punto de referencia más común para determinar la **frecuencia de corte ($f_c$)** en el filtrado espectral de señales.

---

## 1. ¿En qué Consiste la Frecuencia de $-3\text{ dB}$?

La frecuencia de $-3\text{ dB}$ ($\mathbf{f_{-3dB}}$), también conocida como **punto de media potencia** o **frecuencia de esquina**, es la frecuencia a la cual la potencia de salida de un sistema o filtro se ha reducido a la mitad (50%) de su valor máximo en la banda de paso.

### Relación de Amplitud y Potencia

El valor de $-3\text{ dB}$ se deriva de las siguientes relaciones:

1.  **En términos de Potencia ($P$):**
    $$\text{Ganancia en dB} = 10 \cdot \log_{10} \left( \frac{P_{salida}}{P_{máxima}} \right)$$
    Cuando $\frac{P_{salida}}{P_{máxima}} = \frac{1}{2}$ (media potencia):
    $$10 \cdot \log_{10} (0.5) \approx -3.01 \text{ dB}$$

2.  **En términos de Voltaje/Amplitud ($V$):**
    La potencia es proporcional al cuadrado de la amplitud ($P \propto V^2$).
    $$\text{Ganancia en dB} = 20 \cdot \log_{10} \left( \frac{V_{salida}}{V_{máxima}} \right)$$
    Para que la potencia se reduzca a la mitad, la amplitud debe reducirse por el factor $1/\sqrt{2}$ (o $\approx 0.707$).
    $$20 \cdot \log_{10} \left( \frac{1}{\sqrt{2}} \right) \approx -3.01 \text{ dB}$$

Por lo tanto, la frecuencia de $-3\text{ dB}$ es aquella donde la **amplitud de la señal de salida ha caído al $70.7\%$ ($1/\sqrt{2}$) de la amplitud máxima** de la banda de paso.

---

## 2. Relación con la Frecuencia de Corte ($f_c$)

La frecuencia de $-3\text{ dB}$ **es** la frecuencia de corte ($f_c$) por definición en la mayoría de los filtros electrónicos y sistemas de procesamiento de señales.

En el **filtrado espectral de señales**, la frecuencia de corte define el límite entre la **banda de paso** (donde la señal pasa con atenuación mínima) y la **banda de rechazo** (donde la señal es significativamente atenuada).

* **Filtro Paso Bajo:** La $f_c$ es la frecuencia más alta permitida en la banda de paso antes de que la atenuación caiga a $-3\text{ dB}$.
* **Filtro Paso Alto:** La $f_c$ es la frecuencia más baja permitida en la banda de paso antes de que la atenuación caiga a $-3\text{ dB}$.
* **Filtro Paso Banda:** Hay dos frecuencias de corte, $f_{c1}$ (inferior) y $f_{c2}$ (superior), ambas definidas por el punto de $-3\text{ dB}$. El **ancho de banda ($BW$)** del filtro se define como $BW = f_{c2} - f_{c1}$.

En resumen, el punto de **$-3\text{ dB}$** se usa universalmente como un marcador consistente para determinar la **frontera práctica** del filtro, indicando dónde comienza el rechazo significativo de la señal.

```python
import numpy as np
import random
from scipy.spatial.distance import euclidean
from collections import Counter

# =================================================================
# CONFIGURACIÓN DEL DETECTOR
# =================================================================
SR = 22050          # Frecuencia de muestreo (Hz)
DURACION_SEG = 5    # Duración de cada registro (segundos)
N_MUESTRAS = SR * DURACION_SEG
N_REGISTROS = 50    # Registros por género (Total 100)
K = 5               # Número de vecinos más cercanos a considerar (K en KNN)

# =================================================================
# 2. SIMULACIÓN Y CONSTRUCCIÓN DE LA BASE DE DATOS (DB)
# =================================================================
def generar_espectro_simulado(genero, N):
    """
    Simula el espectro de magnitud (sólo la mitad positiva) de una señal de audio.

    Género 1 ('Pop'): Frecuencias bajas/medias dominantes (Pico alrededor de 100-500 Hz).
    Género 2 ('Rock'): Frecuencias medias/altas más energía (Pico alrededor de 500-2000 Hz).
    """
    f_eje = np.fft.fftfreq(N, d=1/SR)[:N//2]

    if genero == 'Pop':
        # Pop: Énfasis en bajas frecuencias (graves y voz central)
        pico_frec = random.uniform(100, 500)
        ancho = 200
    else: # Rock
        # Rock: Énfasis en frecuencias medias-altas (guitarras, percusión)
        pico_frec = random.uniform(500, 2000)
        ancho = 500

    # Función gaussiana para simular el espectro característico
    espectro = np.exp(-0.5 * ((f_eje - pico_frec) / ancho)**2) * random.uniform(0.5, 1.5)

    # Agregar "ruido" espectral para simular variabilidad real
    ruido = np.random.rand(len(espectro)) * 0.1
    return (espectro + ruido) / np.sum(espectro + ruido) # Normalizado

def construir_base_datos(n_registros, N_muestras):
    db = [] # Lista de tuplas: (Espectro_Magnitud, Etiqueta)

    # Género 1: Pop
    for i in range(n_registros):
        espectro = generar_espectro_simulado('Pop', N_muestras)
        db.append((espectro, 'Pop'))

    # Género 2: Rock
    for i in range(n_registros):
        espectro = generar_espectro_simulado('Rock', N_muestras)
        db.append((espectro, 'Rock'))

    random.shuffle(db)
    print(f"Base de datos simulada construida con {len(db)} registros.")
    return db

# =================================================================
# 3. IMPLEMENTACIÓN DEL CLASIFICADOR KNN
# =================================================================

def clasificador_knn_espectral(espectro_nuevo, db, k):
    """
    Clasifica un nuevo registro utilizando KNN basado en la Distancia Euclídea
    entre espectros de magnitud.
    """
    distancias = []

    # 1. Calcular la Distancia Euclídea para cada registro de la DB
    for espectro_db, etiqueta_db in db:
        # Distancia Euclídea entre el espectro de la nueva muestra y el de la DB
        dist = euclidean(espectro_nuevo, espectro_db)
        distancias.append((dist, etiqueta_db))

    # 2. Ordenar por distancia y seleccionar los K vecinos más cercanos
    distancias.sort(key=lambda x: x[0])
    k_vecinos = distancias[:k]

    # 3. Votación de los K vecinos
    etiquetas_vecinos = [etiqueta for dist, etiqueta in k_vecinos]

    # Usar Counter para encontrar la etiqueta más común (la mayoría)
    votos = Counter(etiquetas_vecinos)
    prediccion = votos.most_common(1)[0][0]

    return prediccion, k_vecinos

# =================================================================
# 4. FUNCIÓN DE PRUEBA Y DEMOSTRACIÓN
# =================================================================

def demostrar_detector():
    # Construir la base de datos simulada
    db = construir_base_datos(N_REGISTROS, N_MUESTRAS)

    # Generar una nueva muestra de prueba (simulada)
    genero_prueba = random.choice(['Pop', 'Rock'])
    espectro_prueba = generar_espectro_simulado(genero_prueba, N_MUESTRAS)

    print("\n=============================================")
    print(f"⭐ Nuevo Registro de Prueba (Género Real: {genero_prueba})")
    print("=============================================")

    # Clasificar
    prediccion, vecinos = clasificador_knn_espectral(espectro_prueba, db, K)

    # Mostrar resultados
    print(f"\n✅ Detección completada (K={K})")
    print(f"   -> Género Predicho: {prediccion}")
    print(f"   -> Género Real: {genero_prueba}")

    print("\nTop 5 Vecinos Más Cercanos:")
    for dist, etiqueta in vecinos:
        print(f"   - Distancia: {dist:.4f} | Género DB: {etiqueta}")

# Ejecutar la demostración
if __name__ == "__main__":
    demostrar_detector()
```
